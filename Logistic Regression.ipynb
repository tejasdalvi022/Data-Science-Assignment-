```python
# Drop unnecessary columns
titanic_train = titanic_train.drop(["Name", "Ticket", "Cabin"], axis=1)
```

```python
# Data Preprocessing
warnings.filterwarnings("ignore")
```

```python
# Fill missing values
titanic_train["Embarked"].fillna(titanic_train["Embarked"].mode()[0], inplace=True)
titanic_train["Age"].fillna(titanic_train["Age"].mean(), inplace=True)
```

```python
# Outlier detection and removal using IQR method for Age
Q1 = titanic_train["Age"].quantile(0.25)
Q3 = titanic_train["Age"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
titanic_train = titanic_train[(titanic_train["Age"] >= lower_bound) & (titanic_train["Age"] <= upper_bound)]
```

```python
# Boxplot for Age
plt.boxplot(titanic_train["Age"])
plt.title("Boxplot of Age")
plt.show()
```

```python
# Outlier detection and removal using IQR method for Fare
Q1 = titanic_train["Fare"].quantile(0.25)
Q3 = titanic_train["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
titanic_train = titanic_train[(titanic_train["Fare"] >= lower_bound) & (titanic_train["Fare"] <= upper_bound)]
```

```python
# Boxplot for Fare
plt.boxplot(titanic_train["Fare"])
plt.title("Boxplot of Fare")
plt.show()
```

```python
# Visualizations
# Histogram for Age and Fare
titanic_train[['Age', 'Fare']].hist(bins=30, figsize=(10,4))
plt.suptitle("Distribution of Age and Fare")
plt.show()

# Countplot for Survived vs Pclass
sns.countplot(data=titanic_train, x='Pclass', hue='Survived')
plt.title("Survival count by Passenger Class")
plt.show()

# Pairplot for selected numerical variables
sns.pairplot(titanic_train[['Survived', 'Age', 'Fare', 'SibSp', 'Parch']], hue='Survived')
plt.suptitle("Pairplot of Titanic numerical features", y=1.02)
plt.show()
```

```python
# Label Encoding
label_encoder = LabelEncoder()
titanic_train["Pclass"] = label_encoder.fit_transform(titanic_train["Pclass"])
```

```python
# Initialize label_encoders for Sex and Embarked
label_encoder_sex = LabelEncoder()
titanic_train["Sex"] = label_encoder_sex.fit_transform(titanic_train["Sex"])
```

```python
label_encoder_embarked = LabelEncoder()
titanic_train["Embarked"] = label_encoder_embarked.fit_transform(titanic_train["Embarked"])
```

```python
# Preprocessing Test Set
titanic_test = titanic_test.drop(["Name", "Ticket", "Cabin"], axis=1)
titanic_test["Pclass"] = label_encoder.fit_transform(titanic_test["Pclass"])
titanic_test["Sex"] = label_encoder_sex.transform(titanic_test["Sex"])
titanic_test["Embarked"] = label_encoder_embarked.transform(titanic_test["Embarked"])
titanic_test["Age"].fillna(titanic_test["Age"].mean(), inplace=True)
titanic_test["Fare"].fillna(titanic_test["Fare"].mean(), inplace=True)
```

```python
# Data Visualization
titanic_train["Survived"].value_counts().plot(kind="bar", title="Survival Count")
plt.show()
```

```python
# Logistic Regression Model
X = titanic_train[["Pclass", "Sex", "Age", "SibSp", "Parch", "Fare", "Embarked"]]
y = titanic_train["Survived"]
```

```python
# Split data into train and test sets
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=7)
```

```python
# Fit model on training data
model = LogisticRegression(max_iter=1000)
model.fit(x_train, y_train)
```

```python
# Predictions and evaluation
y_pred = model.predict(x_test)
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: %.2f%%" % (accuracy * 100.0))
print(classification_report(y_test, y_pred))
```

```python
# Confusion Matrix
conf_matrix = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(conf_matrix)
```

```python
# ROC Curve
fpr, tpr, thresholds = roc_curve(y_test, model.predict_proba(x_test)[:, 1])
auc = roc_auc_score(y_test, model.predict_proba(x_test)[:, 1])
```

```python
plt.plot(fpr, tpr, color='red', label='Logit model (AUC = %0.2f)' % auc)
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel('False Positive Rate or [1 - True Negative Rate]')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.title('ROC Curve')
plt.show()
```

```python
# Confusion Matrix
y_pred_train = model.predict(x_train)
sns.heatmap(confusion_matrix(y_train, y_pred_train), annot=True, fmt="d", cmap="Blues")
plt.title("Confusion Matrix")
plt.show()
```

```python
# Interpretation
coefficients = model.coef_[0]
intercept = model.intercept_[0]
print("Coefficients:", coefficients)
print("Intercept:", intercept)
```

Difference Between Precision and Recall Precision measures how many of the predicted positive cases are actually positive, while recall measures how many of the actual positive cases are correctly identified. Precision focuses on prediction accuracy, and recall focuses on coverage of actual positives. High precision means fewer false positives; high recall means fewer false negatives.

What is cross-validation, and why is it important in binary classification? Cross-validation is a technique to evaluate a binary classification model’s ability to generalize by splitting the data into multiple folds, training on some folds, and testing on the remaining one in each iteration. This process repeats so every data point is used for both training and testing. It’s important because it gives a more reliable performance estimate, reduces overfitting risk, and makes better use of limited data.

### Interview Questions:

What is the difference between precision and recall?
Precision and recall are both important metrics in evaluating the performance of a classification model, but they capture different aspects.

Precision is the ratio of correctly predicted positive observations to the total predicted positives. It answers the question: Of all instances that were predicted as positive, how many were actually positive? High precision means that there are very few false positives.

Recall (or sensitivity) is the ratio of correctly predicted positive observations to all the actual positives. It answers the question: Of all the instances that were actually positive, how many did we correctly predict as positive? High recall means that there are very few false negatives.

In short:

Precision is about the accuracy of positive predictions.

Recall is about the coverage of actual positives.
